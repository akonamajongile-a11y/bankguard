# Transactions

```python
import pandas as pd
transactions = pd.read_csv("data/raw/transactions.csv")
transactions.info()
transactions
```

```python
#Clean Transaction Type Categories:

transactions.transaction_type = pd.Series(transactions.transaction_type).str.strip().str.title()

transactions.transaction_type.unique()
```

```python
#Clean Channel Categories:
transactions.channel = pd.Series(transactions.channel).str.strip().str.title()

transactions.channel.unique()
```

```python
transactions.amount =transactions.amount.str.replace("R","")
transactions.amount =transactions.amount.str.replace(" ","")
transactions.amount =transactions.amount.str.replace(",","")
```

```python
# Add feature: Transaction Size (Low, Medium & High)

import pandas as pd
import numpy as np

# Thresholds for each currency
thresholds = {
    "ZAR": {"high": 24999, "medium_low": 7000, "medium_high": 14000},
    "EUR": {"high": 1330, "medium_low": 374.75, "medium_high": 744.00},
    "USD": {"high": 1516, "medium_low": 431.24, "medium_high": 861.24}
}

# Start with all rows as "Low"
transactions["transaction_size"] = "Low"

# ZAR conditions
transactions.loc[
    (transactions["currency"] == "ZAR") & (transactions["amount"] > thresholds["ZAR"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "ZAR") & 
    (transactions["amount"].between(thresholds["ZAR"]["medium_low"], thresholds["ZAR"]["medium_high"])),
    "transaction_size"
] = "Medium"

# EUR conditions
transactions.loc[
    (transactions["currency"] == "EUR") & (transactions["amount"] > thresholds["EUR"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "EUR") & 
    (transactions["amount"].between(thresholds["EUR"]["medium_low"], thresholds["EUR"]["medium_high"])),
    "transaction_size"
] = "Medium"

# USD conditions
transactions.loc[
    (transactions["currency"] == "USD") & (transactions["amount"] > thresholds["USD"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "USD") & 
    (transactions["amount"].between(thresholds["USD"]["medium_low"], thresholds["USD"]["medium_high"])),
    "transaction_size"
] = "Medium"

transactions
```

# Transaction Analytics

```python
# Which transaction type is most common?
transaction_type = (transactions.groupby('transaction_type')['transaction_id'].nunique().sort_values(ascending=False))

print(transaction_type)
print("Most common transaction type:" , transaction_type.idxmax())
```

```python
# Which channels are most frequently used?
channel = (transactions.groupby('channel')['transaction_id'].nunique().sort_values(ascending=False))

print(channel)
top_three = channel.head(3) # top 3 channels
print("Top three most frequently used channels:", top_three)
```

```python
# What are the busiest transaction days?
transaction_days = (transactions.groupby('transaction_days')['transaction_id'].nunique().sort_values(ascending=False))

print(transaction_days)
top_three = transaction_days.head(3) # top 3 channels
print("Top three busiest transaction days:", top_three)
```

```python
# How does transaction behaviour change over time?

import pandas as pd

# Ensure transaction_date is datetime
transactions["transaction_date"] = pd.to_datetime(transactions["transaction_date"], errors="coerce")

# Extract year
transactions["year"] = transactions["transaction_date"].dt.year

# Group by year and count transactions
yearly_totals = transactions.groupby("year")["transaction_id"].count()

print("Total transactions per year:")
print(yearly_totals)
```

```python
g = list(yearly_totals.index.values)[1:]
v = list(yearly_totals.values)[1:]
g

```

```python
import matplotlib.pyplot as plt


# Create Bar Chart
plt.bar(g, v, color='skyblue', edgecolor='black')

# Formatting
plt.title('Transaction Behaviour Chart')
plt.xlabel('Year')
plt.ylabel('Amount')

# Display
plt.show()
```
