```python
# Add feature: Transaction Size (Low, Medium & High)

import pandas as pd
import numpy as np

# Thresholds for each currency
thresholds = {
    "ZAR": {"high": 24999, "medium_low": 7000, "medium_high": 14000},
    "EUR": {"high": 1330, "medium_low": 374.75, "medium_high": 744.00},
    "USD": {"high": 1516, "medium_low": 431.24, "medium_high": 861.24}
}

# Start with all rows as "Low"
transactions["transaction_size"] = "Low"

# ZAR conditions
transactions.loc[
    (transactions["currency"] == "ZAR") & (transactions["amount"] > thresholds["ZAR"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "ZAR") & 
    (transactions["amount"].between(thresholds["ZAR"]["medium_low"], thresholds["ZAR"]["medium_high"])),
    "transaction_size"
] = "Medium"

# EUR conditions
transactions.loc[
    (transactions["currency"] == "EUR") & (transactions["amount"] > thresholds["EUR"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "EUR") & 
    (transactions["amount"].between(thresholds["EUR"]["medium_low"], thresholds["EUR"]["medium_high"])),
    "transaction_size"
] = "Medium"

# USD conditions
transactions.loc[
    (transactions["currency"] == "USD") & (transactions["amount"] > thresholds["USD"]["high"]),
    "transaction_size"
] = "High"

transactions.loc[
    (transactions["currency"] == "USD") & 
    (transactions["amount"].between(thresholds["USD"]["medium_low"], thresholds["USD"]["medium_high"])),
    "transaction_size"
] = "Medium"

transactions
```

# Transaction Analytics

```python
# Which transaction type is most common?
transaction_type = (transactions.groupby('transaction_type')['transaction_id'].nunique().sort_values(ascending=False))

print(transaction_type)
print("Most common transaction type:" , transaction_type.idxmax())
```

```python
g = list(transaction_type.index.values)[0:]
v = list(transaction_type.values)[0:]
g
```

```python
# Horizonal Bar Chart: Transaction Types
import matplotlib.pyplot as plt

# Create Horizontal Bar Chart: Transaction Types vs No. of Transactions
plt.barh(g, v, color="hotpink", edgecolor="black")

# Formatting
plt.title("Transaction Types vs No. of Transactions")
plt.xlabel("No. of Transactions")
plt.ylabel("Transaction Types")

# Display
plt.show()
```

```python
# Which channels are most frequently used?
channel = (transactions.groupby('channel')['transaction_id'].nunique().sort_values(ascending=False))

print(channel)
top_channels = channel.head(3) # top 3 channels
print("Top three most frequently used channels:", top_channels)
```

```python
g = list(top_channels.index.values)[0:]
v = list(top_channels.values)[0:]
g
```

```python
# Pie Chart: Top Channels
import matplotlib.pyplot as plt

plt.pie(
    v,                      # values (number of transactions)
    labels=g,               # labels (channel names)
    autopct="%1.1f%%",      # show percentages on slices
    startangle=90,          # rotate chart for cleaner look
    colors=["hotpink","skyblue","lightgreen"], # optional custom colors
    wedgeprops={"edgecolor":"black"}  # outline slices
)

# Formatting
plt.title("Top Transaction Channels")

# Display
plt.show()
```

# Risk Scoring

```python
import pandas as pd
transactions = pd.read_csv("data/final/transactions_clean_final.csv")
transactions
```

```python
import pandas as pd

# Start with flags as columns in Transactions
transactions["large_transactions_flag"] = (transactions["transaction_size"] == "High").astype(int)
transactions["high_amount_vs_average_flag"] = (transactions["amount"] > transactions["amount"].mean()).astype(int)
transactions["high_risk_merchant_flag"] = transactions["merchant_id"].isin(["M123", "M456"]).astype(int)  
# Example merchant IDs flagged as high risk

# Calculate risk_score
transactions["risk_score"] = (
    transactions["large_transactions_flag"] * 35 +
    transactions["high_amount_vs_average_flag"] * 35 +
    transactions["high_risk_merchant_flag"] * 30
)

# Classify risk based on score
def classify_risk(score):
    if score < 40:
        return "Low"
    elif score < 60:
        return "Medium"
    elif score < 80:
        return "High"
    else:
        return "Very High"

transactions["risk_category"] = transactions["risk_score"].apply(classify_risk)

# Show the first 10 rows with risk_score and risk_category
print(transactions[["transaction_id", "amount", "transaction_size", "risk_score", "risk_category"]].head(10))



    
```
