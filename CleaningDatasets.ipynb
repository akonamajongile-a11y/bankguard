# Transactions

```python
import pandas as pd
transactions = pd.read_csv("data/raw/transactions.csv")
transactions.info()
transactions
```

```python
# Convert data type: Date from string to datetime
transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date'],
    format= 'mixed',
    dayfirst= True,
    errors= 'coerce'
)
transactions
```

```python
# Display Duplicates in dataset
transactions[transactions.duplicated()]
transactions[transactions.duplicated() == True]
```

```python
# Delete Duplicates:
transactions = transactions.drop_duplicates()

transactions[transactions.duplicated() == True]

```

```python
transactions[transactions['merchant_id'].isna()]
```

```python
transactions.merchant_id.fillna("Unknown")
transactions.info()
```

```python
import numpy as np
transactions = transactions.replace(np.nan, "NaN")
```

```python
import numpy as np
transaction_date  = transactions.replace(np.nan, "NaN")
```

```python
transactions.info()
```

```python
#Clean Transaction Type Categories:

transactions.transaction_type = pd.Series(transactions.transaction_type).str.strip().str.title()

transactions.transaction_type.unique()
```

```python
#Clean Channel Categories:
transactions.channel = pd.Series(transactions.channel).str.strip().str.title()

transactions.channel.unique()
```

```python
#Clean Currency Categories:
transactions.currency = transactions.currency.str.strip().str.upper()
transactions.currency = transactions.currency.str.replace('ZAR', "R")


transactions.currency.unique()
```

```python
# transactions[transactions['transaction_date'].isnull() == True]
transactions['transaction_date'] = transactions['transaction_date'].fillna(pd.Timestamp('1900-01-01'))
# transactions
```

```python
transactions.amount =transactions.amount.str.replace("R","")
transactions.amount =transactions.amount.str.replace(" ","")
transactions.amount =transactions.amount.str.replace(",","")
```

```python
transactions.amount = transactions.amount.astype(float)

import numpy as np
transactions['amount'] = transactions['amount'].fillna(np.float64('0.0')) # Replace missing value with '0.0'
```

```python
transactions.to_csv("data/processed/transactions_clean.csv")
```

# Transaction Feature Engineering

```python
# Add a boolean feature: True if amount > 24999

import numpy as np

# Define thresholds where EUR & USD = R24999
thresholds = {
    "ZAR": 24999,
    "EUR": 1330,
    "USD": 1516
}


transactions["is_high_transaction"] = np.where(
    (transactions["currency"] == "ZAR") & (transactions["amount"] > thresholds["ZAR"]), True,
    np.where(
        (transactions["currency"] == "EUR") & (transactions["amount"] > thresholds["EUR"]), True,
        np.where(
            (transactions["currency"] == "USD") & (transactions["amount"] > thresholds["USD"]), True,
            False
        )
    )
)



transactions
```
