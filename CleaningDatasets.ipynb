# Exchange Rates

```python
import pandas as pd
exchange_rates = pd.read_csv("exchange_rates.csv")
exchange_rates.info()
exchange_rates
```

```python
# Convert data type: Date from string to datetime
exchange_rates['date'] = pd.to_datetime(
    exchange_rates['date'],
    format= 'mixed',
    dayfirst= True,
    errors= 'coerce'
)
exchange_rates
```

```python
# Display Duplicates in dataset
exchange_rates[exchange_rates.duplicated() == True]
```

# Transactions

```python
import pandas as pd
transactions = pd.read_csv("transactions.csv")
transactions.info()
transactions
```

```python
# Convert data type: Date from string to datetime
transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date'],
    format= 'mixed',
    dayfirst= True,
    errors= 'coerce'
)
transactions
```

```python
# Display Duplicates in dataset
transactions[transactions.duplicated()]
transactions[transactions.duplicated() == True]
```

```python
# Delete Duplicates:
transactions = transactions.drop_duplicates()

transactions[transactions.duplicated() == True]

```
